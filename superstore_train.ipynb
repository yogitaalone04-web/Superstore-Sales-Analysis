```python
import pandas as pd

df = pd.read_csv('superstore_data.csv')
df.head()
```

```python
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
print(region_sales)
```

```python
category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
print(category_sales)
```

```python
state_sales = df.groupby('State')['Sales'].sum().sort_values(ascending=False).head(10)
print(state_sales)
```

```python
import matplotlib.pyplot as plt

# Region Sales Graph
plt.figure(figsize=(8,5))
region_sales.plot(kind='bar', color='steelblue')
plt.title('Region-wise Sales Analysis')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Category Sales Graph
plt.figure(figsize=(8,5))
category_sales.plot(kind='bar', color='coral')
plt.title('Category-wise Sales Analysis')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Top 10 State Sales Graph
plt.figure(figsize=(10,6))
state_sales.plot(kind='bar', color='green')
plt.title('Top 10 States - Sales Analysis')
plt.xlabel('State')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Dashboard - Teeno graphs ek saath
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Graph 1 - Region
region_sales.plot(kind='bar', ax=axes[0], color='steelblue')
axes[0].set_title('Region-wise Sales')
axes[0].set_xlabel('Region')
axes[0].set_ylabel('Total Sales')
axes[0].tick_params(axis='x', rotation=45)

# Graph 2 - Category
category_sales.plot(kind='bar', ax=axes[1], color='coral')
axes[1].set_title('Category-wise Sales')
axes[1].set_xlabel('Category')
axes[1].set_ylabel('Total Sales')
axes[1].tick_params(axis='x', rotation=45)

# Graph 3 - Top 10 States
state_sales.plot(kind='bar', ax=axes[2], color='green')
axes[2].set_title('Top 10 States - Sales')
axes[2].set_xlabel('State')
axes[2].set_ylabel('Total Sales')
axes[2].tick_params(axis='x', rotation=45)

plt.suptitle('Superstore Sales Dashboard', fontsize=16)
plt.tight_layout()
plt.show()
```
